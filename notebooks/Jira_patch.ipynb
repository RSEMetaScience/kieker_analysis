# Agents

```python
from kieker_analysis.stats.contributor_metrics import get_user_activity, metrics_all
metrics_all = metrics_all(client)
user_activity = get_user_activity(metrics_all)
# drop the columns with only zeros
user_activity = user_activity.loc[:, (user_activity != 0).any()]
user_activity
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
c_matrix = user_activity.corr(method="spearman")
plt.figure(figsize=(30,15))
sns.heatmap(c_matrix, annot=True)
plt.xticks(rotation=30, ha='right')
plt.show()
```
